# Real Data Model Nama Jalan Jakarta

```python
from enum import Enum
from dataclasses import dataclass, field
from typing import List
import random
import difflib

```

```python
class KelasJalan(Enum):
    """Kelas jalan untuk penghitungan pajak reklame."""
    PROTOKOL_A = "PROTOKOL A"
    PROTOKOL_B = "PROTOKOL B"
    PROTOKOL_C = "PROTOKOL C"
    EKONOMI_1 = "EKONOMI 1"
    EKONOMI_2 = "EKONOMI 2"
    EKONOMI_3 = "EKONOMI 3"


class Wilayah(Enum):
    """Wilayah Kota Administrasi DKI Jakarta."""
    JAKARTA_PUSAT = "JAKARTA PUSAT"
    JAKARTA_SELATAN = "JAKARTA SELATAN"
    JAKARTA_TIMUR = "JAKARTA TIMUR"
    JAKARTA_UTARA = "JAKARTA UTARA"
    JAKARTA_BARAT = "JAKARTA BARAT"


class Kecamatan(Enum):
    """Kecamatan di DKI Jakarta."""
    CAKUNG = "CAKUNG"
    CEMPAKA_PUTIH = "CEMPAKA PUTIH"
    CENGKARENG = "CENGKARENG"
    CILANDAK = "CILANDAK"
    CILINCING = "CILINCING"
    CIPAYUNG = "CIPAYUNG"
    CIRACAS = "CIRACAS"
    DUREN_SAWIT = "DUREN SAWIT"
    GAMBIR = "GAMBIR"
    GROGOL_PETAMBURAN = "GROGOL PETAMBURAN"
    JAGAKARSA = "JAGAKARSA"
    JATINEGARA = "JATINEGARA"
    JOHAR_BARU = "JOHAR BARU"
    KALIDERES = "KALIDERES"
    KEBAYORAN_BARU = "KEBAYORAN BARU"
    KEBAYORAN_LAMA = "KEBAYORAN LAMA"
    KEBON_JERUK = "KEBON JERUK"
    KELAPA_GADING = "KELAPA GADING"
    KEMAYORAN = "KEMAYORAN"
    KEMBANGAN = "KEMBANGAN"
    KEPULAUAN_SERIBU = "KEPULAUAN SERIBU"
    KOJA = "KOJA"
    KRAMAT_JATI = "KRAMAT JATI"
    MAKASAR = "MAKASAR"
    MAMPANG_PRAPATAN = "MAMPANG PRAPATAN"
    MATRAMAN = "MATRAMAN"
    MENTENG = "MENTENG"
    PADEMANGAN = "PADEMANGAN"
    PALMERAH = "PALMERAH"
    PANCORAN = "PANCORAN"
    PASAR_MINGGU = "PASAR MINGGU"
    PASAR_REBO = "PASAR REBO"
    PENJARINGAN = "PENJARINGAN"
    PESANGGRAHAN = "PESANGGRAHAN"
    PULOGADUNG = "PULOGADUNG"
    SAWAH_BESAR = "SAWAH BESAR"
    SENEN = "SENEN"
    SETIABUDI = "SETIABUDI"
    TAMAN_SARI = "TAMAN SARI"
    TAMBORA = "TAMBORA"
    TANAH_ABANG = "TANAH ABANG"
    TANJUNG_PRIOK = "TANJUNG PRIOK"
    TEBET = "TEBET"


@dataclass
class JalanEntry:
    """Single entry representing a street and its road class for advertisement tax calculation."""
    no: int
    wilayah: Wilayah
    kecamatan: Kecamatan
    nama_jalan: str
    kelas_jalan: KelasJalan
```

```python
#HELPER FUNCTIONS

def get_jalan_by_kelas(kelas: KelasJalan) -> List[JalanEntry]:
    """Get all streets with a specific road class."""
    return [j for j in DAFTAR_JALAN if j.kelas_jalan == kelas]


def get_jalan_by_kecamatan(kecamatan: Kecamatan) -> List[JalanEntry]:
    """Get all streets in a specific sub-district."""
    return [j for j in DAFTAR_JALAN if j.kecamatan == kecamatan]


def get_jalan_by_wilayah(wilayah: Wilayah) -> List[JalanEntry]:
    """Get all streets in a specific administrative city area."""
    return [j for j in DAFTAR_JALAN if j.wilayah == wilayah]


def search_jalan(keyword: str) -> List[JalanEntry]:
    """Search streets by name keyword (case-insensitive)."""
    keyword_upper = keyword.upper()
    return [j for j in DAFTAR_JALAN if keyword_upper in j.nama_jalan.upper()]


if __name__ == "__main__":
    print(f"Total entries: {len(DAFTAR_JALAN)}")
    print(f"\nEntries by Wilayah:")
    for w in Wilayah:
        count = len(get_jalan_by_wilayah(w))
        print(f"  {w.value}: {count}")
    print(f"\nEntries by Kelas Jalan:")
    for k in KelasJalan:
        count = len(get_jalan_by_kelas(k))
        print(f"  {k.value}: {count}")
```

```python
# Set how many data points to generate
number_of_rec_tocreate = 20
```

```python
variationformats = [
    "Mohon diantar ke {kodepos} - {housenumber} - {kelurahan} - {wilayah} - {rtrw} - {kecamatan} - {dkijakarta} - {namajalan} - {complexname} dgn senyum.",
    "Kirim ke: {namajalan} {housenumber} -  {complexname}, {kelurahan}, - {wilayah} - {rtrw} {kecamatan}, {kodepos} - {dkijakarta} .",
    "Tujuan: {complexname} {namajalan} {housenumber} , {kelurahan}, - {wilayah} - {rtrw}  -  {kecamatan}, {kodepos} - {dkijakarta} .",
    "{complexname} {namajalan} {housenumber} , {kelurahan}, - {wilayah} - {rtrw}  -  {kecamatan},  - {dkijakarta} - {kodepos}",
    "{namajalan} / {complexname} / {housenumber} / {kecamatan}, {kelurahan}, - {wilayah} - {rtrw} / {dkijakarta} - {kodepos}"
]
```

```python
# Pick most detailed data: nama jalan
random_entry = random.choice(DAFTAR_JALAN)
for i in range(number_of_rec_tocreate):
    random_entry = random.choice(DAFTAR_JALAN)
    # print("Kecamatan: "+ random_entry.kecamatan.value)
    # Enrich with data from location_data
    
    ratio = 0;
    while ratio < 0.8:
        a_location = random.choice(location_data)
        ratio = difflib.SequenceMatcher(None, 
                                        random_entry.kecamatan.value.upper(), 
                                        a_location.Kecamatan.upper()).ratio()
        
    random_entry.Kelurahan = a_location.Kelurahan
    random_entry.KodelPos = a_location.KodePos


    alamatstring = random.choice(variationformats).format(kodepos=random_entry.KodelPos, 
                                                        housenumber="No. 15",
                                                        kelurahan=random_entry.Kelurahan,
                                                          wilayah=random_entry.wilayah.value,
                                                          rtrw="RT02/05",
                                                          kecamatan=random_entry.kecamatan.value,
                                                          dkijakarta="DKI Jakarta",
                                                          namajalan=random_entry.nama_jalan,
                                                          complexname="Komp. Deperin")

    # print(random_entry)                                                                                                                                                                         
    # print(alamatstring)
```
